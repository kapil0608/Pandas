```python
def job_clean(row):
    j=row['job']
    d=row['Dname']
    if d=='Accounts Receiveable':
        d= 'Accounts Receivable'
    if j=='CHief Information Officer':
        j='Chief Information Officer'
    if j != 'Auditor' and j != 'Dairy Person':
        if d in j:
            return j.replace(d,'').strip().strip(',').strip(', s')
        else:
            return j.removeprefix('Exec Assistant').strip(',')
    else:
        return j

    # here to place the 'CHief Information Officer' we can also use 
    # df['job']=df['job'].replace('CHief Information Officer','Chief Information Office)
```

```python
df['job']=df.apply(job_clean,axis=1)
```
