# Melt Function

used for apply aggregate function such as min,max,count,sum,mean

for converting into row

syntax= pd.melt(Data, id_vars= , value_vars= , var_name= , value_name=)

id_vars = columns to remain same

value_vars = The columns to change into rows

var_name = New name for the old columns

value_name = column name for the values

```python
pd.melt(df,value_vars=['hotstar','netflix'])
```

```python
pd.melt(df,id_vars=['Movies','rating'],value_vars=['hotstar','netflix'])
```

```python
pd.melt(df,id_vars=['Movies','rating'],value_vars=['hotstar','netflix'], var_name="Platform")
```

```python
df1=pd.melt(df,id_vars=['Movies','rating'],value_vars=['hotstar','netflix'], var_name="Platform",value_name="Avilability")

df1       # new table that we created
```

```python
df1=df1.loc[df1['Avilability']==1]
df1
```

```python
df1=df1.reset_index(drop=True)
df1
```

```python
df1=df1.drop(columns=['Avilability'])
df1
```
