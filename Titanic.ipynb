```python
def comp_type(row):                         # Companion types
    f=row['fam_cnt']
    p=row['no_of_psgs']
    if p==1:
        return 'Solo'
    elif f==0 and p>1:
        return "Friends"
    else:
        return "Family"
```

```python
### syntax---> apply(user_def(), axis=1)                    # apply function works like a for loop

df['companion type']=df.apply(comp_type , axis=1)                                # axis =1 that means it apply row by row   
```

```python
Royal=['the Countess','Jonkheer','Sir','Don','Dona']
Common=['Mrs','Mr',"Lady",'Ms','Mme','Mile','Miss',"Master"]
Military=['Major','Col','Capt']
other_proffesion=['Dr','Rev']
```

```python
def social_status(row):
    t=row['title']
    if t in Royal:
        return 'Royal'
    elif t in Common:
        return 'Common'
    elif t in Military:
        return 'Military'
    else:
        return 'other_proffesion'

df['social_status']=df.apply(social_status,axis=1)
    
```

2. Bivarient Plot ----> we use barplot(barplot is used in categorical data and numerical data)

   x-axis=Categorical data
   
   y-axis=Numerical data

```python
plt.figure(figsize=(3,4))
b=sns.barplot(data=df,x='gender',y='fare', errorbar=None,estimator='mean')
b.bar_label(b.containers[0],fmt='%.1f')
plt.title('Avg fare by gender wise')
plt.yticks(range(0,20,2))
plt.show()  # Remove Unwanted data
```

***Observation***

Females pays higher avg fare than the male

```python
plt.figure(figsize=(3,4))
b=sns.barplot(data=df,x='gender',y='fare', errorbar=None,estimator='sum',)
b.bar_label(b.containers[0],fmt='%.1f')
plt.title('Total fare by gender wise')
plt.yticks(range(0,14000,2000))
plt.show()  # Remove Unwanted data
```

***Observation***

Males has higher total fare than the Females

```python
plt.figure(figsize=(5,4))
b=sns.barplot(data=df,x='age category',y='fare', errorbar=None,estimator='sum',)
b.bar_label(b.containers[0],fmt='%.1f')
plt.title('Total fare by gender wise')
plt.yticks(range(0,14000,2000))
plt.show()  # Remove Unwanted data
```

```python
plt.figure(figsize=(5,4))
b=sns.barplot(data=df, x='age category', y='fare', errorbar=None,hue='gender', palette=['cyan','teal'],estimator="mean")
for i in b.containers:
    b.bar_label(i,fmt='%.1f')
plt.yticks(range(0,40,5))
plt.legend(loc='center',bbox_to_anchor=(0.5,0.9),ncols=2)
plt.show()
```

***Observation***

Average fare increase by the age Category, among all the category female pays higher than the male
